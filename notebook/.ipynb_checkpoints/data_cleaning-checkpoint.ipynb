# 2. Data Cleansing

```python
# remove leading/ trailing whitespace
col_names = {col: col.strip() for col in data.columns} # create a dict with key of original name and value of modified name

# equal to
count_affected = 0
col_names = {}
for col in data.columns:
    if col != col.strip():
        count_affected += 1
    col_names[col] = col.strip()

print(f"Number of column affected: {count_affected}")
data.rename(columns = col_names, inplace = True)
```

## Remove duplicates and infinity

```python
duplicates = data.duplicated()
duplicate_count = duplicates.sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# remove duplicate rows
data = data.drop_duplicates(keep='first')
del duplicates
data.shape
```

```python
# remove duplicate columns
identical_columns = {} # dict key = column name, values = identical columns
columns = data.columns
list_control = columns.copy().tolist()

for col1 in columns:
    for col2 in columns:
        if col1 != col2:
            if data[col1].equals(data[col2]):
                # create a new key = value entry in the dict
                if (col1 not in identical_columns) and (col1 in list_control):
                    identical_columns[col1] = [col2]
                    list_control.remove(col2)
                # update the key value enrty by appending a new column to the list
                elif (col1 in identical_columns) and (col1 in list_control):
                    identical_columns[col1].append(col2)
                    list_control.remove(col2)

# print the result
if identical_columns:
    print("Identical columns found:")
    for key, value in identical_columns.items():
        print(f"{key} is identical to {value}")
else:
    print("No identical columns found")
```

```python
# removing columns with duplicate values
for key, value in identical_columns.items(): #identical_columns.items() print tuples
    data.drop(columns=value, inplace=True) # inplace=True modifies DataFrame direclty instead of a copy

print(data.columns)
data.shape
```

```python
# checking for infinite values
numeric_columns = data.select_dtypes(include = np.number).columns # select columns with numerical values
contains_infinite = np.isinf(data[numeric_columns]).sum() # a dataframe with column name as index, sum of infinity as values
print(contains_infinite[contains_infinite > 0]) # print series with morethan 0 of infinite values
```

```python
# identify highly correlated features
threshold = 0.85
high_correlation = np.where(np.abs(corr_matrix) > threshold) # return row and column indices
high_correlation = [(corr_matrix.index[x], corr_matrix.columns[y], corr_matrix.iloc[x, y]) 
                 for x, y in zip(*high_correlation) if x != y and x < y] # create tuple with row, col, corr value
high_correlation[:5] # print 5 first values
```

```python
# printing pairs with high correlation and isolating the near perfect multicollinearity
high_multicollinearity = []

print("Items with high correlation:")
for item in high_correlation:
    print(f"[{item[0]}]--[{item[1]}]: {item[2].round(4)}")
    if item[2] >= 0.95:
        high_multicollinearity.append(item)
```

```python
# plotting pairs with high_multicollinearity

n_plots = len(high_multicollinearity)
n_cols = 4
n_rows = (n_plots + n_cols - 1) // n_cols # finding minimal no.of rows

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, n_rows * 4)) # figsize(width, height in inches)
axes = axes.flatten()

for i, item in enumerate(high_multicollinearity):
    feature_x = item[0]
    feature_y = item[1]
    corr_value = item[2]

    # Scatter plot
    sns.scatterplot(x=data[feature_x], y=data[feature_y], ax=axes[i])
    axes[i].set_title(f'{feature_x} vs {feature_y} (Corr={corr_value:.2f})', fontsize=8)
    axes[i].set_xlabel(feature_x, fontsize=8)
    axes[i].set_ylabel(feature_y, fontsize=8)

# Hide any unused subplots
for j in range(len(high_multicollinearity), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

## 3.3. Outliers

Identifying features with significant deviation (outlier) is crucial in improving model accuracy by preventing skewed result

```python
def calculate_outlier_percentage(df):
    outlier_percentages = {}

    for column in df.columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1 # inter wuartile range
    
        lower_bound = Q1 - 1.5 * IQR # lower bound
        upper_bound = Q3 + 1.5 * IQR # upper bound
    
        outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
        outlier_percentage = len(outliers) / len(df) * 100
    
        outlier_percentages[column] = outlier_percentage
    
    return outlier_percentages
```

```python
outlier_percentages = calculate_outlier_percentage(data[numeric_features])
```

```python
outlier_percentages
```

```python
# convert to DataFrame for easier manipulation
outliers_df = pd.DataFrame.from_dict(outlier_percentages, orient='index', columns=['Outlier_Percentage'])

# define the threshold
threshold = 10

# identify features with high percentage of outliers
high_outlier_features = outliers_df[outliers_df['Outlier_Percentage'] > threshold]
high_outlier_features
```

```python
# plot outlier percentages and highlight features above threshold

outliers_df.sort_values(by='Outlier_Percentage', ascending=False).plot(
    kind='bar', 
    legend=False, 
    figsize=(20, 5)
) # using matplotlib wrapper builtin to DataFrames to setup x and y axis
plt.axhline(y=threshold, color='r', linestyle='--', label=f"{threshold}% Threshold")
plt.xlabel('Features')
plt.ylabel('Percentage of oultiers')
plt.title('Perncentage of ourliers for each feature with threshold')
plt.legend()
plt.show()
```

**Class Imbalance**

The previous analysis show a clear class imbalance, which should be taken into consideration while working with the CIC-IDS-2017, as highlighted by Panigrahi and Borah (2018).

**Observations**:
1. Dominance of the 'BENIGN' class/ Normal Traffic – 83.1% of the entire dataset
2. Higher types with lower counts – Even DoS, the second most relevant class, shows low counts
3. Rare classes even after regrouping, such as Bots, Web Attacks and Brute Force

**Potential implications**:
* Model Bias: Models can become biased towards the majority class, leading to high accuracy but poor recall and precision for minority classes.
* Performance Metrics: In an imbalanced dataset, accuracy is often not a sufficient metric to evaluate model performance. F1-score, precision, recall, or ROC-AUC are better suited for assessing the model's ability to classify different classes effectively.

**Ideas to address the issue during ML training**:
* Class weights to penalize mistakes on minority classes
* Ensemble Methods can improve overall predictions (Random Forests, XGBoost)
* Cross-validation with representative folds
* Partial SMOTE to help balancing the distribution
* Undersampling the majority class

# 5. Featrue Importance

```python
# 3. Variance analysis (df, numeric_features, target_col='Attack Type') – Levene -> dict with results

from scipy import stats

def analyze_variance_homogeneity(df, numeric_features, target_col='Attack Type'):  
    """  
    Analyze the homogeneity of variances using Levene's test.  

    Parameters:  
    -----------  
    df : pandas.DataFrame  
        Input DataFrame containing the dataset.  
    numeric_features : list  
        List of column names containing numerical features to analyze. 
    target_col : str, optional (default='Attack Type')
        Name of the target column

    Returns:  
    --------  
    dict  
        Dictionary containing the results of Levene's test for each feature.  
    """  
    
    results_levene = {}  

    groups = []

    # loops over numeric feature
    for feature in numeric_features:  
        # splits df into sub-df(group) based on 'Attack Type'
        for name, group in df.groupby(target_col): # name is values of Attack Type, group is the df
            feature_dropna = group[feature].dropna() # drop NaN values from the feature
            if not feature_dropna.empty: # if feature is not empty, add the val
                val = feature_dropna.values
                # if feature is not empty, feature is more than zero, apppend to groups
                if len(val) > 0 and np.any(val != 0) and np.var(val) > 0:
                    groups.append(val)

        # Check if there are at least two groups with valid data  
        if len(groups) < 2:  
            print(f"Not enough valid groups to perform Levene's test for feature: {feature}")  
            continue  # Skip this feature if not enough valid groups  

        # Perform Levene's Test  
        stat_levene, p_value_levene = stats.levene(*groups)  
        results_levene[feature] = {'Statistic': stat_levene, 'p-value': p_value_levene}  

    return results_levene
```

```python
variance_result = analyze_variance_homogeneity(data, numeric_features)
```
